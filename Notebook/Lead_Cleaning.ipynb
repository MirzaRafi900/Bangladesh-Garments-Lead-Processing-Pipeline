# Library Load

```python
import pandas as pd
import numpy as np
```

# Load Dataset

```python
df = pd.read_excel(
    "../Dataset/2k Garments and buying data.xlsx"
)

print(df.shape)

df.head()
```

# Missing & duplicated Values

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
lead_df = df[
    [
        "Bcode",
        "Company Name",
        "Company type",
        "ContactName",
        "Designation",
        "OfficeAdd1",
        "OfficeAdd2",
        "OfficeSubZone",
        "OfficeZone",
        "Email1",
        "Mobile",
        "IndustryType"
        
    ]
]
print(lead_df.shape)
```

```python
lead_df = lead_df.dropna(
    subset = ["Email1","Mobile"],
    how="all"
)

lead_df = lead_df.fillna(
    "Not Available"
)

print(lead_df.shape)
```

```python
lead_df = lead_df.drop_duplicates()
print(lead_df.shape)
```

```python
lead_df.head(500)
```

```python
lead_df["Mobile"] = (
    lead_df["Mobile"].astype(str)
)
```

```python
lead_df.to_excel(
    "../Cleaned_Dataset/Lemon_500_Leads.xlsx",
    index=False
)
```
